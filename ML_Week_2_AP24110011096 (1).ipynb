```python
#Column Selection

import pandas as pd
info = {'one' : pd.Series([1, 2, 3, 4, 5, 6], index=['a', 'b', 'c', 'd', 'e', 'f']),
'two' : pd.Series([1, 2, 3, 4, 5, 6, 7, 8], index=['a', 'b', +
                                                'c', 'd', 'e', 'f', 'g', 'h'])}
d1 = pd.DataFrame(info)
print (d1 ['one'])
print(d1['two'])
```

```python
#Column Addition

import pandas as pd
info = {'one' : pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e']),
'two' : pd.Series([1, 2, 3, 4, 5, 6], index=['a', 'b', 'c', 'd', 'e', 'f'])}
df = pd.DataFrame(info)
print(df)
# Add a new column to an existing DataFrame object
print ("Add new column by passing series")
df['three']=pd.Series([20,40,60],index=['a','b','c'])
print (df)
print ("Add new column using existing DataFrame columns")
df['four']=df['one']+df['three']
print (df)
```

```python
#Column Deletion:

import pandas as pd

info = {'one' : pd.Series([1, 2], index= ['a', 'b']),
'two' : pd.Series([1, 2, 3], index=['a', 'b', 'c'])}

df = pd.DataFrame(info)
print ("The DataFrame:")
print (df)

# using del function
print ("Delete the first column:")
del df['one']
print (df)
# using pop function
print ("Delete the another column:")
df.pop('two')
print (df)
```

```python
#Selection by Label

import pandas as pd
info = {'one' : pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e']),
'two' : pd.Series([1, 2, 3, 4, 5, 6], index=['a', 'b', 'c', 'd', 'e', 'f'])}

df = pd.DataFrame(info)
print("Data frame:")
print(df)
print("b index elements are:")
print (df.loc['b'])
print("e index elements are:")
print(df.loc['e'])
```

```python
# Import the Pandas library
import pandas as pd

# Read the Iris dataset from the CSV file
df = pd.read_csv('D:/SRM_CSE_Class_Work/AY_24_25_SRM-CSE/AY_24_25-OddSem/ML_C303/Materials/Lab_Experiments/iris.csv')

# Display the complete dataset
print(df)

# Group the data based on the 'species' column
# Calculate the mean (average) of the numerical columns for each species
mean_df = df.groupby('species').mean(numeric_only=True)

# Display the mean values for each species
print("Mean values for each species:")
print(mean_df)

# Group the data based on the 'species' column
# Calculate the sum of the numerical columns for each species
sum_df = df.groupby('species').sum(numeric_only=True)

# Display the sum values for each species
print("\nSum of values for each species:")
print(sum_df)
```
